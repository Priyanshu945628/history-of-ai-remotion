### Step 3: Render the Video (Live Streaming Logs)

```python
import subprocess
import os

os.makedirs('out', exist_ok=True)
print('🚀 Starting Remotion 3D Render (11,171 frames)...')

# Execute with unbuffered live streaming output
process = subprocess.Popen(
    ['npx', 'remotion', 'render', 'HistoryOfAI', 'out/history-of-ai.mp4', '--gl=swangle', '--log=verbose'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    universal_newlines=True,
    bufsize=1
)

for line in process.stdout:
    print(line, end='', flush=True)

process.wait()
if process.returncode == 0:
    print('\n✅ Rendering completed successfully! File saved at out/history-of-ai.mp4')
    print(f'Size: {os.path.getsize("out/history-of-ai.mp4") / (1024*1024):.2f} MB')
else:
    print(f'\n❌ Render exited with code {process.returncode}')
```
